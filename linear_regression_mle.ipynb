```python
#liner regression
import matplotlib.pyplot as plt
# Imports the pyplot module from Matplotlib. It is used to create graphs and plots.

from scipy import stats
# Imports the statistics module from SciPy. We will use stats.linregress() to calculate linear regression.


# Data
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

# Perform simple linear regression
slope, intercept, r, p, std_err = stats.linregress(x, y)

# slope     → slope of the regression line
# intercept → value of y when x = 0
# r         → correlation coefficient
# p         → p-value
# std_err   → standard error of the slope

# Function for predicted values
def myfunc(x):
    return slope * x + intercept
# Build model
mymodel = list(map(myfunc, x))


# map() applies myfunc() to every value in x.
# For example:
# myfunc(5)
# myfunc(7)
# myfunc(8)
# ...
# The predicted y values are stored in mymodel.
# list() converts the map object into a list.


# Plot
#STEP 5: Plot the original data points
plt.scatter(x, y)
# STEP 6: Plot the regression line
plt.plot(x, mymodel)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Linear Regression Example")
plt.show()
```

```python
from google.colab import files
import pandas as pd
uploaded = files.upload()
BosData = pd.read_csv('BostonHousing.csv')
print(BosData)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
import pandas as pd

# Load dataset

# Select features and target
x = BosData.iloc[:, :11]   # first 11 columns as features
y = BosData.iloc[:, 13]    # column 13 as target

# Train-test split
x_train, x_test, y_train, y_test = train_test_split( x, y, test_size=0.2, random_state=11
)

# Train model
reg = LinearRegression()
reg.fit(x_train, y_train)

# Predictions
y_test_predict = reg.predict(x_test)
# Metrics
mae = mean_absolute_error(y_test, y_test_predict)
mse = mean_squared_error(y_test, y_test_predict)
rmse = np.sqrt(mean_squared_error(y_test, y_test_predict))
r2 = r2_score(y_test, y_test_predict)

# Output o Test Data
print("Test MAE:",mae)
print("MSE:",mse)
print("Test RMSE:", rmse)
print("Test R2 score:", r2)
```

The regression model achieves an R² score of approximately 0.61, meaning that it explains about 61% of the variation in house prices. The MAE of 3.74 indicates that the average prediction differs from the actual value by about 3.74 units, while the RMSE of 5.81 indicates that larger prediction errors have a noticeable effect on the model's performance.
